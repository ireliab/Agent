```python
import os
from dataclasses import dataclass
from typing import Literal

from deepagents import create_deep_agent
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from tavily import TavilyClient

# Secrets live in .env (gitignored); see .env.example for the keys it needs.
# This also picks up LANGSMITH_TRACING / LANGSMITH_API_KEY / LANGSMITH_PROJECT.
load_dotenv()


@dataclass
class Context:
    user_id: str
    api_key: str


tavily_client = TavilyClient(api_key=os.environ["TAVILY_API_KEY"])


def internet_search(
    query: str,
    max_results: int = 5,
    topic: Literal["general", "news", "finance"] = "general",
    include_raw_content: bool = False,
):
    """Run a web search"""
    return tavily_client.search(
        query,
        max_results=max_results,
        include_raw_content=include_raw_content,
        topic=topic,
    )
```

```python
# System prompt to steer the agent to be an expert researcher
research_instructions = """
You are a helpful assistant
"""

llm = ChatOpenAI(
    model=os.environ["MODEL_NAME"],
    api_key=os.environ["MODEL_API_KEY"],
    base_url=os.environ["MODEL_BASE_URL"],
)

agent = create_deep_agent(
    model=llm,
    tools=[internet_search],
    system_prompt=research_instructions,
)
```

```python
result = agent.invoke({"messages": [{"role": "user", "content": "What is langgraph? Use tools when needed for detail research"}]})
```

```python
for m in result["messages"]:
    m.pretty_print()
```
